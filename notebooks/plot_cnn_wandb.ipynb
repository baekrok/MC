```python
import wandb
import numpy as np
import matplotlib.pyplot as plt
```

```python
from collections import defaultdict


api = wandb.Api(timeout=100)
resnet_cifar10_runs = api.runs(f"backrockqwer/matrix_completion_rank_resnet_cifar10")
# vgg_cifar10_runs = api.runs(f"backrockqwer/matrix_completion_rank_vgg_cifar10")
resnet_cifar100_runs = api.runs(f"backrockqwer/matrix_completion_rank_resnet_cifar100")
# vgg_cifar100_runs = api.runs(f"backrockqwer/matrix_completion_rank_vgg_cifar100")

resnet_cifar10_rank = defaultdict(list)
resnet_cifar10_train_acc = defaultdict(list)
resnet_cifar10_test_acc = defaultdict(list)
for run in resnet_cifar10_runs:
    rank = run.history()["rank/avg_eff"].values  # summary에서 값 가져오기
    train_acc = run.history()['train/acc_top1'].values
    test_acc = run.history()['val/acc_top1'].values
    resnet_cifar10_rank[run.name].append(rank)
    resnet_cifar10_train_acc[run.name].append(train_acc)
    resnet_cifar10_test_acc[run.name].append(test_acc)

resnet_cifar100_rank = defaultdict(list)
resnet_cifar100_train_acc = defaultdict(list)
resnet_cifar100_test_acc = defaultdict(list)
for run in resnet_cifar100_runs:
    rank = run.history()["rank/avg_eff"].values  # summary에서 값 가져오기
    train_acc = run.history()['train/acc_top1'].values
    test_acc = run.history()['val/acc_top1'].values
    resnet_cifar100_rank[run.name].append(rank)
    resnet_cifar100_train_acc[run.name].append(train_acc)
    resnet_cifar100_test_acc[run.name].append(test_acc)


# vgg_cifar10_rank = defaultdict(list)
# vgg_cifar10_train_acc = defaultdict(list)
# vgg_cifar10_test_acc = defaultdict(list)
# for run in vgg_cifar10_runs:
#     rank = run.history()["rank/avg_eff"].values  # summary에서 값 가져오기
#     train_acc = run.history()['train/acc_top1'].values
#     test_acc = run.history()['val/acc_top1'].values
#     vgg_cifar10_rank[run.name].append(rank)
#     vgg_cifar10_train_acc[run.name].append(train_acc)
#     vgg_cifar10_test_acc[run.name].append(test_acc)

# vgg_cifar100_rank = defaultdict(list)
# vgg_cifar100_train_acc = defaultdict(list)
# vgg_cifar100_test_acc = defaultdict(list)
# for run in vgg_cifar100_runs:
#     rank = run.history()["rank/avg_eff"].values  # summary에서 값 가져오기
#     train_acc = run.history()['train/acc_top1'].values
#     test_acc = run.history()['val/acc_top1'].values
#     vgg_cifar100_rank[run.name].append(rank)
#     vgg_cifar100_train_acc[run.name].append(train_acc)
#     vgg_cifar100_test_acc[run.name].append(test_acc)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_with_ci(ax, arr, label):
    """
    arr: shape (num_runs, num_epochs)
    """
    arr = np.array(arr)
    mean = arr.mean(axis=0)
    stderr = arr.std(axis=0) / np.sqrt(arr.shape[0])
    ci = 1.96 * stderr   # 95% CI
    x = np.arange(len(mean))
    ax.plot(x, mean, label=label)
    ax.fill_between(x, mean-ci, mean+ci, alpha=0.2)

fig, ax = plt.subplots(1, 3, figsize=(9, 2.7))

models = [
    ('arch-resnet18_opt-rmsprop_data-cifar100_lr-0.001_bsz-128_wd-1e-05', 'ResNet-18'),
    ('arch-resnet34_opt-rmsprop_data-cifar100_lr-0.001_bsz-128_wd-1e-05', 'ResNet-34'),
    ('arch-resnet50_opt-rmsprop_data-cifar100_lr-0.001_bsz-128_wd-1e-05', 'ResNet-50'),
    ('arch-resnet101_opt-rmsprop_data-cifar100_lr-0.001_bsz-128_wd-1e-05', 'ResNet-101'),
]

# Train Accuracy
for key, name in models:
    plot_with_ci(ax[0], resnet_cifar100_train_acc[key], name)

# Test Accuracy
for key, name in models:
    plot_with_ci(ax[1], resnet_cifar100_test_acc[key], name)

# Average Rank
for key, name in models:
    plot_with_ci(ax[2], resnet_cifar100_rank[key], name)

ax[0].set_ylabel('Train Accuracy')
ax[1].set_ylabel('Test Accuracy')
ax[2].set_ylabel('Average Effective Rank')

for a in ax:
    a.set_xlabel('Epochs')

ax[0].set_ylim(61, 101)
ax[1].set_ylim(53, 67.8)

plt.legend()
plt.tight_layout()
plt.savefig('resnet_cifar100_rmsprop.pdf')
```
